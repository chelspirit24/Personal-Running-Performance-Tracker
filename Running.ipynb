```python
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('export.csv')
df
```

```python
df.columns.unique()
```

```python
df_filtered = df[['workoutActivityType','type','value','unit','duration', 'durationUnit', 'startDate','endDate']]
df_filtered
```

```python
df[df['type'].str.contains('run', case=False, na=False)]['type'].unique()
df[df['workoutActivityType'].str.contains('run',case=False, na=False)]['workoutActivityType'].unique()
```

```python
runs = df_filtered[df_filtered['workoutActivityType'] == 'HKWorkoutActivityTypeRunning']
runs
```

```python
runs.isna().sum()
runs_filtered = runs.drop(columns = ['type', 'value', 'unit'])
runs_filtered
```

```python
distance = df_filtered[df_filtered['type'] == 'HKQuantityTypeIdentifierDistanceWalkingRunning'].copy()
distance
```

```python
distance.isna().sum()
distance_filtered = distance.drop(columns = ['workoutActivityType', 'duration', 'durationUnit'])
distance_filtered
```

```python
runs_filtered['startDate'] = pd.to_datetime(runs_filtered['startDate'], errors='coerce')
runs_filtered['endDate'] = pd.to_datetime(runs_filtered['endDate'], errors='coerce')
distance_filtered['startDate'] = pd.to_datetime(distance_filtered['startDate'], errors='coerce')
distance_filtered['value'] = pd.to_numeric(distance_filtered['value'], errors='coerce')
```

```python
def calculate_distance(run, dist_df):
    in_time_window = (dist_df['startDate'] >= run['startDate']) & (dist_df['startDate'] <= run['endDate'])
    
    return dist_df.loc[in_time_window, 'value'].sum()

runs_filtered['distance'] = runs_filtered.apply(lambda row: calculate_distance(row, distance_filtered), axis=1)
runs_filtered 
```

```python
latest_date = runs_filtered['startDate'].max()
one_year_ago = latest_date - pd.DateOffset(years = 1)
runs_one_year = runs_filtered[runs_filtered['startDate'] >= one_year_ago].copy()
runs_one_year = runs_one_year.sort_values('startDate')
runs_one_year
```

```python
runs_one_year['pace'] = runs_one_year['duration'] / runs_one_year['distance']
```

```python
running_one_year_data = runs_one_year[['startDate','distance', 'pace']]
running_one_year_data
```

```python
running_one_year_data = running_one_year_data[
    (running_one_year_data['pace'] > 6) & (running_one_year_data['pace'] < 15) & (running_one_year_data['distance'] > 1)
]
running_one_year_data
```

```python
distance_plot = running_one_year_data.plot(x='startDate', y='distance', figsize=(10, 5), title='Running Distance Over Time')
plt.xlabel ('Date')
plt.ylabel('Distance in Miles')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
pace_plot = running_one_year_data.plot(x='startDate', y='pace', figsize=(10, 5), title='Running Pace Over Time', color = 'purple')
plt.xlabel ('Date')
plt.ylabel('Pace in Mins per Mile')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
